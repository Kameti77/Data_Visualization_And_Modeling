```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

```python
df = pd.read_csv('Employee_productivity.csv')

df.head(15)
```

## Hours worked

```python
sns.regplot(x='HoursWorked', y='ProductivityScore', data=df)
plt.title('Productivity by Hours Worked')
plt.ylabel('Productivity Score')
plt.show()
```

```python
X = sm.add_constant(df['HoursWorked'])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

## Years at company

```python
sns.regplot(x='YearsAtCompany', y='ProductivityScore', data=df)
plt.title('Productivity by Years at Company')
plt.ylabel('Productivity Score')
plt.show()
```

```python
X = sm.add_constant(df['YearsAtCompany'])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

## Focus score

```python
sns.regplot(x='FocusScore', y='ProductivityScore', data=df)
plt.title('Productivity by Focus Score')
plt.ylabel('Productivity Score')
plt.show()
```

```python
X = sm.add_constant(df['FocusScore'])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

## Which feature has the strongest linear relationship with productivity?
Employees who are better able to focus are dramatically more productive. Neither how long they've been at the company nor how many hours they work has any meaningful impact on productivity scores alone.

## Task 2: Building a Multi-Variable Model

### Step 1: Add TeamMeetingsPerWeek

```python
X = sm.add_constant(df[['FocusScore', 'TeamMeetingsPerWeek']])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

###   Step 2: Add HoursWorked

```python
X = sm.add_constant(df[['FocusScore', 'TeamMeetingsPerWeek', 'HoursWorked']])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

### Step 3: Add the categorical columns

#### Desk location

```python
df['DeskLocation'].value_counts()
```

```python
df['DeskLocation_enc'] = df['DeskLocation'].map({'Open Plan': 0, 'Private Office': 1, 'Remote': 2})
```

#### Department

```python
df['Department'].value_counts()
```

```python
df['Department_enc'] = df['Department'].map({'HR': 0, 'Engineering': 1, 'Sales': 2, 'Marketing': 3})
```

#### Add desk location and department to the model

```python
X = sm.add_constant(df[['FocusScore', 'TeamMeetingsPerWeek', 'HoursWorked', 
                          'DeskLocation_enc']])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

```python
X = sm.add_constant(df[['FocusScore', 'TeamMeetingsPerWeek', 'HoursWorked', 
                          'DeskLocation_enc', 'Department_enc']])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

### Observation
Based on the p-values from the full model, only FocusScore (p = 0.000) and DeskLocation (p = 0.038) were statistically significant predictors of productivity (p < 0.05). TeamMeetingsPerWeek, HoursWorked, and Department all had p-values well above 0.05, meaning their effect on productivity could simply be due to chance. We therefore build a cleaner final model using only the two significant variables, which avoids overfitting by not including variables that don't meaningfully contribute to predicting productivity.

```python
X = sm.add_constant(df[['FocusScore', 'DeskLocation_enc']])
y = df['ProductivityScore']
model = sm.OLS(y, X).fit()
model.summary()
```

## Summary & Recommendations
After analyzing all the available factors, it is clear that FocusScore is by far the most important driver of employee productivity, explaining 68% of the variation in productivity scores on its own. This means that an employee's ability to concentrate and stay focused is the single biggest determinant of how productive they are, far more than any other factor in the dataset.

Interestingly, two of the most common complaints at this company turned out to be completely unsupported by the data. Neither how many hours someone works nor how long they have been at the company had any meaningful relationship with productivity. This means that tenured employees are not inherently lazier than new ones, and new employees are not inherently less focused, both groups were simply wrong about each other.
Desk location was the only other statistically significant factor, though its effect was small. Contrary to what employees believed, working in an open plan area did not hurt productivity compared to having a private office or working remotely.

Meetings, department, and hours worked were all statistically insignificant, meaning none of them reliably predict whether an employee will be productive or not.

## Recommendations
The company should focus its efforts on creating an environment where employees can focus, this means reducing unnecessary interruptions, limiting meetings that don't have a clear purpose, and addressing whatever is causing low morale, since stress and dissatisfaction are well known enemies of focus. Judging employees by how many hours they work or how long they have been at the company is not only unfair but also not backed by the data. Productivity at this company is fundamentally a focus problem, not a hours or experience problem.
